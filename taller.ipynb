## **Punto 2 – Operaciones Matemáticas entre Columnas**

En esta sección se realizarán operaciones matemáticas entre las columnas del DataFrame con el fin de generar nueva información y analizar el comportamiento de las ventas y costos de los productos tecnológicos.


* **Crear una nueva columna llamada Total_Venta = Precio × Cantidad**

Se calculará el valor total de venta de cada producto multiplicando el precio por la cantidad disponible.

```python
#Elimina espacios invisibles en nombres de columnas (MUY importante)
df.columns = df.columns.str.strip()

# Limpia la columna PRECIO eliminando símbolos como "$" y puntos, y convierte los valores a tipo numérico decimal (float)
df["PRECIO"] = df["PRECIO"].replace(r'[\$,\.]', '', regex=True).astype(float)

#Convierte la columna CANTIDAD a tipo entero para poder realizar operaciones matemáticas correctamente
df["CANTIDAD"] = df["CANTIDAD"].astype(int)

 # Crea una nueva columna llamada TOTAL_VENTA multiplicando el precio por la cantidad
df["TOTAL_VENTA"] = df["PRECIO"] * df["CANTIDAD"]

# Mostrar primeras filas para verificar
df.head()
```

**Explicación**

Se realizó la limpieza y preparación de los datos para su correcto análisis. En primer lugar, se eliminaron los espacios invisibles en los nombres de las columnas utilizando df.columns.str.strip() para evitar errores al manipular los datos.

Luego, se limpió la columna PRECIO, eliminando símbolos como “$” y separadores de miles, convirtiendo los valores a tipo numérico decimal (float) para permitir operaciones matemáticas.

Posteriormente, la columna CANTIDAD fue convertida a tipo entero, asegurando que los datos fueran numéricos y aptos para cálculos.

Finalmente, se creó la columna TOTAL_VENTA, la cual resulta de multiplicar el precio por la cantidad de cada producto. Esta operación permite calcular el valor total de ventas generado por cada producto dentro del conjunto de datos. Para validar los cambios, se visualizaron las primeras filas del DataFrame.


* **Crear una nueva columna llamada Costo_Total = Total_Venta + Costo_Envio**

Se calculará el costo total de cada producto sumando el valor de venta y el costo de envío.

```python
# Limpiar columna COSTO DE ENVIO
df["COSTO DE ENVIO"] = df["COSTO DE ENVIO"].replace(r'[\$,\.]', '', regex=True).astype(float)

# Crear una nueva columna llamada COSTO_TOTAL
# Se suma el total de ventas y el costo de envío de cada producto
df["COSTO_TOTAL"] = df["TOTAL_VENTA"] + df["COSTO DE ENVIO"]

# Mostrar las primeras filas del DataFrame con la nueva columna creada
df.head()
```

**Explicación**

Se realizó la limpieza de la columna COSTO DE ENVÍO, eliminando símbolos como “$” y separadores de miles, y convirtiendo los valores a tipo numérico decimal para permitir operaciones matemáticas.

Posteriormente, se creó la columna COSTO_TOTAL, la cual resulta de la suma entre TOTAL_VENTA y COSTO DE ENVÍO. Esta operación permite obtener el valor total asociado a cada producto, considerando tanto el valor de venta como el costo del envío.

Finalmente, se visualizaron las primeras filas del DataFrame para verificar que la nueva columna fue creada correctamente


* **Calcular el promedio del Total_Venta**

Se calculará el promedio de las ventas totales registradas en el conjunto de datos, con el fin de conocer el comportamiento general de los valores de venta de los productos tecnológicos.

```python
# Calcular el promedio de la columna TOTAL_VENTA
promedio_ventas = df["TOTAL_VENTA"].mean()

# Mostrar el promedio de ventas obtenido
print("El promedio de ventas es:", promedio_ventas)
```

**Explicación**

La función **mean()** permite calcular el promedio de los valores almacenados en la columna **TOTAL_VENTA**.

El resultado obtenido representa el valor promedio de las ventas registradas para los productos tecnológicos del conjunto de datos.

* **Calcular qué porcentaje representa cada producto respecto al total general de ventas**

Se determinará la participación porcentual de cada producto en relación con el total general de ventas, permitiendo analizar cuáles tienen mayor relevancia dentro del conjunto de datos.

```python
# Calcular el total general de ventas sumando todos los valores de TOTAL_VENTA
TOTAL_GENERAL = df["TOTAL_VENTA"].sum()

# Crear una nueva columna que representa el porcentaje de participación de cada producto
 # respecto al total general de ventas
df["PORCENTAJES_VENTAS"] = (df["TOTAL_VENTA"] / TOTAL_GENERAL) * 100

# Mostrar las primeras filas del DataFrame con la nueva columna creada
df.head()
```

**Explicación**

Primero se calcula el total general de ventas sumando todos los valores de la columna **TOTAL_VENTA**.

Luego, se crea la columna **PORCENTAJES_VENTAS**, la cual se obtiene dividiendo la venta de cada producto entre el total general y multiplicando el resultado por 100.

De esta manera, se determina el porcentaje de participación de cada producto dentro del total de ventas, lo que permite identificar su nivel de relevancia en el conjunto de datos.

## **Punto 4 – Estadística Descriptiva**

En esta sección se aplican funciones de estadística descriptiva con el fin de analizar el comportamiento de los datos numéricos del conjunto de información. A través de estas medidas se obtiene una visión general sobre la tendencia central, la dispersión y los valores extremos de las variables más importantes del DataFrame.


* **Calcular la media de la columna PRECIO**

Se calcularán medidas de tendencia central en la columna PRECIO con el fin de analizar el comportamiento general de los valores, identificando el promedio, el valor central y el más frecuente.

```python
# Calcular la media de la columna PRECIO
media_precio = df["PRECIO"].mean()

# Mostrar el resultado del promedio de los precios
print("La media del precio es:", media_precio)
```

**Explicación**

La media corresponde al promedio de los valores de la columna **PRECIO**.

Esta medida permite obtener un valor representativo del conjunto de precios de los productos tecnológicos, facilitando el análisis general de su comportamiento.


* **Calcular la mediana de la columna PRECIO**

Se calculará el valor central de los precios ordenados, lo que permite identificar el punto medio del conjunto de datos.

```python
# Calcular la mediana de la columna PRECIO
mediana_precio = df["PRECIO"].median()

# Mostrar el valor de la mediana de los precios
print("La mediana del precio es:", mediana_precio)
```

**Explicación**

La mediana corresponde al valor central de la columna `PRECIO` una vez que los datos se encuentran ordenados.

Esta medida divide el conjunto de datos en dos partes iguales, permitiendo analizar la tendencia central de los precios sin verse afectada por valores extremos.


* **Calcular la moda de la columna PRECIO**

Se identificará el valor de precio que más se repite dentro del conjunto de datos, lo que permite conocer el valor más frecuente en la columna.

```python
# Mostrar el valor de la moda (valor más frecuente) de la columna PRECIO
# Calcular la moda de la columna PRECIO
moda_precio = df["PRECIO"].mode()

# Mostrar la moda
print("La moda del precio es:", moda_precio[0])
```

**Explicación**

Esta instrucción muestra el valor de la moda de la columna **PRECIO**, es decir, el precio que más se repite dentro del conjunto de datos.

Se utiliza **moda_precio[0]** para obtener el primer valor de la moda en caso de que existan varios resultados.


* **Calcular la desviación estándar de la columna CANTIDAD**

Se determinará la variabilidad de los datos en la columna CANTIDAD, lo que permite conocer qué tan dispersos están los valores respecto al promedio.

```python
# Quitar espacios en los nombres de las columnas
df.columns = df.columns.str.strip()

# Calcular la desviación estándar de la columna CANTIDAD
desviacion = df["CANTIDAD"].std()

# Mostrar el resultado
print("La desviación estándar de CANTIDAD es:", desviacion)
```

**Explicación**

La desviación estándar calculada fue de 6.09, indicando el grado de variación o dispersión de los datos respecto a la media. Esta medida estadística es útil para identificar qué tan homogéneos o dispersos son los valores analizados

La desviación estándar es una medida estadística que indica cuánto se alejan los valores de la columna CANTIDAD respecto a su promedio.

Si la desviación estándar es alta, significa que las cantidades varían mucho entre sí: algunos productos tienen valores muy grandes y otros muy pequeños.

Si la desviación estándar es baja, las cantidades están más uniformes y cercanas al promedio, mostrando poca variabilidad.

En otras palabras, la desviación estándar nos ayuda a entender si los datos son consistentes o si presentan una gran dispersión dentro del conjunto analizado.

## **Punto 5 – Filtrado**

En esta sección se aplican técnicas de filtrado y ordenamiento de datos con el fin de extraer información específica del DataFrame. Esto permite seleccionar subconjuntos de datos según condiciones establecidas y organizar la información de manera más clara para su análisis.


* **Mostrar únicamente los productos de la categoría Tecnología**

Se filtrará el DataFrame para seleccionar únicamente los registros pertenecientes a la categoría Tecnología, con el fin de analizar este grupo de productos de manera específica.

```python
# Normalizar la columna CATEGORIA
df["CATEGORIA"] = df["CATEGORIA"].str.strip().str.upper()

# Filtrar productos de TECNOLOGIA
productos_tecnologia = df[df["CATEGORIA"] == "TECNOLOGIA"]

# Mostrar resultado
display(productos_tecnologia)
```

**Explicación**

Primero se eliminan los espacios en blanco de los nombres de las columnas utilizando **str.strip()**, con el fin de evitar errores al momento de acceder a ellas.

Luego, se realiza un filtro sobre la columna **CATEGORIA** para seleccionar únicamente los registros correspondientes a la categoría **TECNOLOGIA**.

Finalmente, se muestran los datos filtrados en forma de tabla, permitiendo visualizar solo los productos pertenecientes a dicha categoría.


* **Ordenar los productos de mayor a menor según Total_Venta**

Se organizarán los productos en orden descendente según la columna TOTAL_VENTA, lo que permite identificar cuáles generan mayores niveles de ventas dentro del conjunto de datos.

```python

# Ordenar el DataFrame de mayor a menor según la columna TOTAL_VENTA
productos_ordenados = df.sort_values(by="TOTAL_VENTA", ascending=False)

# Mostrar los productos ordenados
display(productos_ordenados)
```

**Explicación**

La función **sort_values()** permite ordenar los datos del DataFrame según los valores de una columna específica.

En este caso, los productos se ordenan de mayor a menor según la columna **TOTAL_VENTA**, lo que facilita identificar cuáles generan mayores niveles de ventas dentro del conjunto de datos.
